```python
#Step 2: Load and inspect the data
```

```python
df = pd.read_csv("cookie_cats.csv")
print(df.shape)
df.head()
```

```python
# Step 3: Duplicates and Missing values
```

```python
df.info()
print("Missing values:", df.isnull().sum().sum())
print("Duplicate users:", df['userid'].duplicated().sum())
```

```python
#Step 4: Check the group split
```

```python
print(df['version'].value_counts())
print(df['version'].value_counts(normalize=True).round(3))
```

```python
#Step 5: Explore game rounds and find the outlier
```

```python
df['sum_gamerounds'].describe()
```

```python
sns.boxplot(x=df['sum_gamerounds'])
plt.title("Game rounds – before removing outlier")
plt.show()

print("Players with 0 rounds:", (df['sum_gamerounds'] == 0).sum())
```

#Step 6: Remove the outlier

```python
df = df[df['sum_gamerounds'] < df['sum_gamerounds'].max()]
print(df.shape)
```

```python
#Step 7: Calculate retention by group
```

```python
summary = (df.groupby('version')[['retention_1', 'retention_7']].mean() * 100).round(2)
summary
```

```python
summary.plot(kind='bar')
plt.ylabel("Retention (%)")
plt.title("Day-1 and Day-7 retention by version")
plt.xticks(rotation=0)
plt.show()
```

```python
#Step 8: Write the hypotheses (Markdown cell)
# H0: no difference in D7 retention between gate_30 and gate_40.
# H1: there is a difference.
# α = 0.05. If p < 0.05, reject H0.
```

```python
# Step 9: Run the significance test (two-proportion z-test)
```

```python
def ab_test(metric):
    grp = df.groupby('version')[metric].agg(['sum', 'count'])
    z, p = proportions_ztest(grp['sum'].values, grp['count'].values)
    print(f"{metric}: z = {z:.3f}, p-value = {p:.4f}")

ab_test('retention_1')
ab_test('retention_7')
```

```python
for m in ['retention_1', 'retention_7']:
    table = pd.crosstab(df['version'], df[m])
    chi2, p, dof, exp = stats.chi2_contingency(table)
    print(f"{m}: chi-square p-value = {p:.4f}")
```

```python
#Step 10: Confidence intervals
```

```python
for v in ['gate_30', 'gate_40']:
    g = df[df['version'] == v]['retention_7']
    low, high = proportion_confint(g.sum(), len(g), alpha=0.05)
    print(f"{v}: D7 = {g.mean()*100:.2f}%  (95% CI: {low*100:.2f}% – {high*100:.2f}%)")
```

```python
#Step 11: Bootstrap (bonus, impresses interviewers)
```

```python
boot = []
for i in range(1000):
    s = df.sample(frac=1, replace=True)
    m = s.groupby('version')['retention_7'].mean()
    boot.append((m['gate_30'] - m['gate_40']) * 100)

boot = pd.Series(boot)
boot.plot(kind='hist', bins=30, title="Bootstrap: D7 difference (gate_30 − gate_40), % points")
plt.show()
print("Probability gate_30 is better:", (boot > 0).mean())
```

```python
#Step 12: Conclusion (final Markdown cell)
Conclusion: Moving the gate from level 30 to level 40 lowered 7-day retention from 19.02% to 18.20%, and this drop is statistically significant (z-test p ≈ 0.0016, with non-overlapping 95% confidence intervals). The 1-day retention difference (44.82% vs 44.23%) was not significant (p ≈ 0.07).

Recommendation: Keep the gate at level 30. An earlier gate seems to help players stay engaged longer.

Business impact: with thousands of new installs every day, a drop of about 0.8 percentage points in D7 retention means hundreds fewer active players each week, and less in-game revenue.
```
